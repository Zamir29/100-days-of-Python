### NaN (Not A Number) check
Use the `isna()` combined with `any()` that works with boolean data with `axis=1` that goes in row-wise direction (`axis=0` goes column-wise).

```python
df.isna().tail()
```

```python
df.isna().any()
```

Create a mask with `.any()` and `axis=1`. While `df.isna().any(axis=1)` gives back a **Series of booleans** with the same length as the dataframe, it means it can be used as a mask for the dataframe.  
It means we can have a table where only the row that has at least (any) columns with value NaN will appear.

```python
df[df.isna().any(axis=1)]
```

## Edit rows
Use `dropna()` to drop the last row that simply refers to the source and it is not usable. 

```python
clean_df = df.dropna()
# Check the tail to confirm the row was dropped
clean_df.tail()
```

## Accessing data

### Find College Major with the Highest **Starting Salary**

#### Locate the highest value for the Starting Salary and the index
The `.max()` alone returns for a Series just a value, in this case `np.float64(74300.0)`.  
Instead `.idxmax()` return the index of the row with the highest value

```python
clean_df['Starting Median Salary'].max()
```

```python
max_salary_starting = clean_df['Starting Median Salary'].idxmax()
print(max_salary_starting)
```

#### Locate the name of the Undergraduate Major
Use `.loc` (location) to check the property of a particular row

```python
clean_df['Undergraduate Major'].loc[max_salary_starting]
```

```python
# With one square brackets it shows a series, with two it forces to a 2D dimension table (list of series) and as such shows a table
clean_df.loc[[max_salary_starting]]
```
