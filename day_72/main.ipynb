### NaN (Not A Number) check
Use the `isna()` combined with `any()` that works with boolean data with `axis=1` that goes in row-wise direction (`axis=0` goes column-wise).

```python
df.isna().tail()
```

```python
df.isna().any()
```

Create a mask with `.any()` and `axis=1`. While `df.isna().any(axis=1)` gives back a **Series of booleans** with the same length as the dataframe, it means it can be used as a mask for the dataframe.  
It means we can have a table where only the row that has at least (any) columns with value NaN will appear.

```python
df[df.isna().any(axis=1)]
```

## Sorting Values & Adding Columns

### Lowest Risk Majors

Comparison is the spread between the 10<sup>th</sup> percentiles and the 90<sup>th</sup>

```python
spread_col = abs(clean_df['Mid-Career 10th Percentile Salary'] - clean_df['Mid-Career 90th Percentile Salary'])
```

Insert the Series `spread_col` between the 10th and 90th column.

```python
clean_df.insert(4, "Spread", spread_col)
clean_df.head()
```

Sort the values to check the lowest to the highest spread

```python
low_risk = clean_df.sort_values(by='Spread', ascending=True)
low_risk[['Undergraduate Major', 'Spread']].head()
```

### Greatest Opportunities

Degrees with the highest spread

```python
low_risk = clean_df.sort_values(by='Spread', ascending=False)
low_risk[['Undergraduate Major', 'Spread']].head()
```

Top 5 degrees with the highest values in the 90th percentile

```python
clean_df[['Undergraduate Major', 'Mid-Career 90th Percentile Salary']].sort_values(by='Mid-Career 90th Percentile Salary', ascending=False).head()
```

Top 5 salaries after graduation

```python
clean_df[['Undergraduate Major', 'Mid-Career Median Salary']].sort_values(by='Mid-Career Median Salary', ascending=False).head()
```

## Grouping and Pivoting

Count number of degree for each category

```python
clean_df.groupby('Group')[['Undergraduate Major']].count()
```

The mean salary for each **Group**

```python
pd.options.display.float_format = '{:,.2f}'.format
clean_df.groupby('Group')[['Starting Median Salary','Mid-Career Median Salary']].mean()
```
